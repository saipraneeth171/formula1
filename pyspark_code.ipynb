```python
struct type --> over all schema as collection i.e blueprint
struct field --> individual columns/ fields inside the struct type 
```

```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, DoubleType

circuits_schema = StructType(fields=
                             [
                             StructField("circuitId", IntegerType(), False),
                             StructField("circuitRef", StringType(), True),
                             StructField("name", StringType(), True),
                             StructField("location", StringType(), True),
                             StructField("country", StringType(), True),
                             StructField("lat", DoubleType(), True),
                             StructField("lng", DoubleType(), True),
                             StructField("alt", IntegerType(), True),
                             StructField("url", StringType(), True)
                             ])
```

```python
df = spark.read\
    .schema(circuits_schema)\
    .option("header", True)\
    .csv("abfss://demofiles@formula1adls.dfs.core.windows.net/circuits.csv")
df.display()
```

```python
from pyspark.sql.functions import col
# df.select("circuitId", "circuitRef").show()
# df.select(col("circuitId"),col("circuitRef")).display()
df.select("*").show()
```

```python
#add new column to the data frame 
from pyspark.sql.functions import lit,current_timestamp
df_sample= df.withColumn("new_column", lit(1))
df_sample= df_sample.withColumnRenamed("lat", "latitude")
df_sample = df_sample.withColumn("updated_timestamp", current_timestamp())\
    .withColumn("demo",lit("pyspark"))
df_sample.display()
```

```python
df_sample=df_sample.drop("new_column", "updated_timestamp")
df_sample.display()
```

```python
df_sample.write.mode("overwrite").format("delta").saveAsTable("formula1_dev.bronze.circuits_sample")
```
