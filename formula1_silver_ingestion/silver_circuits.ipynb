```python
from pyspark.sql.functions import initcap, upper,lower,trim,ltrim,rtrim
df2= df.withColumn("circuitRef1", initcap(col("circuitRef")))\
    .withColumn("circuitRef_upper", upper(col("circuitRef")))\
        .withColumn("circuitRer_lower", lower(col("circuitRef_upper")))\
            .withColumn("circuitRef_trim", trim(col("circuitRef")))\
                .withColumn("circuitRef_ltrim", ltrim(col("circuitRef")))\
                    .withColumn("circuitRef_rtrim", rtrim(col("circuitRef")))\
                        .display()
```

```python
# latitude -->  < 100 -->low 
# <500 -->medium 
# null --> unknown
from pyspark.sql.functions import col, when,lit
df_circuits = df.withColumn("circuit_label", when(col("latitude") < 100, "Low")
                                            .when(col("latitude") < 500, "Medium")
                                            .otherwise("Unknown"))
df12=df_circuits.withColumn("new", lit(1))
```
