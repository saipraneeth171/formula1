```python
df_clean= df.dropDuplicates(["driver_id", "c2"])
df_clean.display()
```

```python
df_clean.printSchema()
```

```python
df_null =df_clean.fillna({"number": 0}) # 
df_null.display() 
```

```python
# df_null.display()
from pyspark.sql.functions import concat,lit, col,concat_ws, year, month, dayofmonth
df_drivers = df_null.withColumn("forname", col("name.forename"))\
    .withColumn("surname", col("name.surname"))\
        .withColumn("full_name", concat(col("forname"), lit(" "), col("surname")))\
            .withColumn("fuull_nam1", concat_ws(" ", col("name.forename"), col("name.surname")))\
                .withColumn("year", year(col("dob")))\
                    .withColumn("month", month(col("dob")))\
                        .withColumn("day", dayofmonth(col("dob"))).display()
```
