```python
df_clean= df.dropDuplicates(["driver_id"])
df_clean.display()
```

```python
df_clean.printSchema()
```

```python
df_null =df_clean.fillna({"number": 0}) # 
df_null.display()
```

concat & date functions

```python
# df_null.display()
from pyspark.sql.functions import concat,lit, col,concat_ws, year, month, dayofmonth,current_timestamp,date_add,date_sub,current_date,dayofweek,hour,minute
df_drivers = df_null.withColumn("forname", col("name.forename"))\
    .withColumn("surname", col("name.surname"))\
        .withColumn("full_name", concat(col("forname"), lit(" "), col("surname")))\
            .withColumn("fuull_nam1", concat_ws(" ", col("name.forename"), col("name.surname")))\
                .withColumn("year", year(col("dob")))\
                    .withColumn("month", month(col("dob")))\
                        .withColumn("day", dayofmonth(col("dob")))\
                            .withColumn("silver_timestamp", current_timestamp())\
                                .withColumn("add_timestamp", date_add(current_date(), 10))\
                                    .withColumn("sub_timestamp", date_sub(current_date(), 360))\
                                        .withColumn("weekofday", dayofweek(current_date()))\
                                            .withColumn("hour", hour(current_timestamp()))\
                                                .withColumn("minute", minute(current_timestamp()))
```

```python
df_drivers.write.mode("overwrite").format("delta").saveAsTable("formula1_dev.silver.drivers")
```
