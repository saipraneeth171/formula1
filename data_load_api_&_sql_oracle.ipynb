```python
import requests 
#step 1 call the API URI
url = "https://jsonplaceholder.typicode.com/posts"
# api_key = ""
response = requests.get(url) # url hit

if response.status_code == 200: # http 200 ok 
    data = response.json()
    #step2 :api response ni data frame lo store chesukovali
    df = spark.createDataFrame(data)
    df.count()
else:
    print(f"failed to fecth data: {response.status_code}")
```

```python
# jdbc  to connect to the oracle server to the extract the data to the adls gen2 
#step1 
jdbc_url_oracle = "jdbc::oracle:thin:@//<server>:1521/<service_name>"

connection_properties = {
    "user": "<username>",
    "password": "<password>",
    "driver": "oracle.jdbc.driver.OracleDriver"
}
```

```python
#READ  DATA/TABLE FROM SQL SERVER
source_table = "formula1_dev.customers"

df = spark.read.jdbc(
    url = jdbc_url,
    table = source_table,
    property=connection_properties
)
df.display()
```
